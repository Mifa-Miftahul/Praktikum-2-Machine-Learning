E. REFLEKSI
  1. -Supervised Learning adalah metode machine learning yang menggunakan data yang sudah memiliki label.
  -unsupervised adalah menggunakan data yang belum memiliki label untuk mencari pola sendiri.
  -reinforcement adalah pembelajarannya berinteraksi dengan lingkungan dan umpan baliknya bisa berupa reward dan penalty/punishment
  2. pembagian data latih dan data uji penting untuk mengetahui apakah si model bisa bekerja dengan baik pada data yang belum pernah dilihat
  3. alur kerja ML disebut iteratif itu karena prosesnya dilakukan berulang ulang. hasil evaluasi dari model yang sudah dibuat dan di uji yaitu untuk melakukan perbaikan seperti memperbaiki data, sampai performanya sesuai.
